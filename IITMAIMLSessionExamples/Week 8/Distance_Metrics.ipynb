## Distance Metrics in Unsupervised Learning

In unsupervised learning, distance metrics play a crucial role in determining the similarity or dissimilarity between data points. The choice of distance metric can significantly affect the performance of algorithms like **K-Means**, **Hierarchical Clustering**, and **Nearest Neighbors**.

### What is a Distance Metric?

A **distance metric** is a mathematical formula that defines the distance between two data points in a feature space. Different distance metrics measure this distance in different ways, and the choice of the metric depends on the structure of the data and the task at hand. Some common distance metrics include:

1. **Euclidean Distance**: The straight-line distance between two points in space.
2. **Manhattan Distance**: The distance between two points measured along axes at right angles (like navigating a grid).
3. **Cosine Distance**: Measures the cosine of the angle between two vectors, used when the direction is more important than the magnitude.

### Why are Distance Metrics Important?

Distance metrics allow algorithms to group similar points or identify outliers. For example, in **K-Means clustering**, the distance metric helps to assign points to their nearest cluster center. Similarly, in **anomaly detection**, distance metrics help identify points that deviate significantly from the rest of the data.

### How Do We Measure the Distance?
Now, let's calculate the Euclidean, Manhattan, and Cosine distances between these two points to understand how each metric measures distance differently. These metrics will help us better understand the similarities or dissimilarities between data points in our analysis

### **Euclidean Distance**

**Euclidean distance** is the most commonly used distance metric. It measures the straight-line distance between two points in a Euclidean space, making it intuitive and useful for understanding spatial relationships between points.

### **Formula**

The formula for Euclidean distance between two points \( A(x_1, y_1) \) and \( B(x_2, y_2) \) in a 2D space is:

$$
d(A, B) = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}
$$

This formula calculates the distance by determining the difference in the \( x \)-coordinates and \( y \)-coordinates, squaring these differences, summing them, and finally taking the square root to obtain the straight-line distance.

For higher-dimensional spaces, the Euclidean distance formula extends to:

$$
d(A, B) = \sqrt{\sum_{i=1}^{n} (B_i - A_i)^2}
$$

Where \( n \) is the number of dimensions, and \( A_i \) and \( B_i \) are the coordinates of points \( A \) and \( B \) in each dimension.

### **Example**

Let’s calculate the Euclidean distance between the points \( A(2, 3) \) and \( B(7, 1) \).

Using the formula:

$$
d(A, B) = \sqrt{(7 - 2)^2 + (1 - 3)^2}
$$

First, calculate the differences:

- \( (7 - 2) = 5 \)
- \( (1 - 3) = -2 \)

Next, square the differences:

- \( 5^2 = 25 \)
- \( (-2)^2 = 4 \)

Now sum the squares:

$$
25 + 4 = 29
$$

Finally, take the square root:

$$
d(A, B) = \sqrt{29} \approx 5.39
$$

Thus, the Euclidean distance between points \( A(2, 3) \) and \( B(7, 1) \) is approximately **5.39**.


```
import numpy as np

# Define the coordinates of points A and B
A = np.array([2, 3])
B = np.array([7, 1])

# Calculate Euclidean distance
euclidean_distance = np.sqrt(np.sum((B - A)**2))

# Print the result
print(f'Euclidean Distance between A and B: {euclidean_distance:.2f}')
```

### **Manhattan Distance**

**Manhattan distance**, also known as **Taxicab distance** or **L1 distance**, measures the distance between two points by summing the absolute differences of their coordinates. It is called "Manhattan distance" because it mimics the way distances are measured in a grid-like path, such as navigating through the streets of Manhattan, where you can only move vertically or horizontally, not diagonally.

### **Formula**

For two points \( A(x_1, y_1) \) and \( B(x_2, y_2) \) in a 2D plane, the Manhattan distance is calculated as:

$$
d_{\text{Manhattan}}(A, B) = |x_2 - x_1| + |y_2 - y_1|
$$

In an \( n \)-dimensional space, the formula generalizes to:

$$
d_{\text{Manhattan}}(A, B) = \sum_{i=1}^{n} |B_i - A_i|
$$

Where \( |B_i - A_i| \) represents the absolute difference between the coordinates of points \( A \) and \( B \) along each dimension.

### **Example**

Let’s calculate the Manhattan distance between the points \( A(2, 3) \) and \( B(7, 1) \).

Using the formula:

$$
d_{\text{Manhattan}}(A, B) = |7 - 2| + |1 - 3|
$$

First, calculate the absolute differences:

- \( |7 - 2| = 5 \)
- \( |1 - 3| = 2 \)

Now sum the absolute differences:

$$
d_{\text{Manhattan}}(A, B) = 5 + 2 = 7
$$

Thus, the Manhattan distance between points \( A(2, 3) \) and \( B(7, 1) \) is **7**.



```

import numpy as np

# Define the coordinates of points A and B
A = np.array([2, 3])
B = np.array([7, 1])

# Calculate Manhattan distance
manhattan_distance = np.sum(np.abs(B - A))

# Print the result
print(f'Manhattan Distance between A and B: {manhattan_distance}')
```


This implementation computes the **Manhattan distance** between the points \( A(2, 3) \) and \( B(7, 1) \) as **7**, matching the manual calculation.



### **Cosine Distance**

In data science, **Cosine Distance** is used to measure the dissimilarity between two non-zero vectors. Unlike Euclidean or Manhattan distance, Cosine Distance focuses on the **angle** between the vectors, making it a popular metric for high-dimensional spaces like text analysis or document similarity tasks.

### **Formula**

The formula for **Cosine Similarity** between two vectors \( A \) and \( B \) is:

$$
\text{Cosine Similarity} = \frac{A \cdot B}{||A|| \times ||B||}
$$

Where:
- \( A \cdot B \) is the **dot product** of vectors \( A \) and \( B \),
- \( ||A|| \) is the **magnitude** (or length) of vector \( A \), calculated as \( \sqrt{\sum A_i^2} \),
- \( ||B|| \) is the **magnitude** of vector \( B \), calculated similarly.

To convert Cosine Similarity into **Cosine Distance**, we use:

$$
\text{Cosine Distance} = 1 - \text{Cosine Similarity}
$$

### **Example**

Let's calculate the **Cosine Distance** between two points \( A(2, 3) \) and \( B(7, 1) \).

1. **Step 1: Compute the dot product** \( A \cdot B \):

$$
A \cdot B = (2 \times 7) + (3 \times 1) = 14 + 3 = 17
$$

2. **Step 2: Compute the magnitudes** of \( A \) and \( B \):

$$
||A|| = \sqrt{(2^2) + (3^2)} = \sqrt{4 + 9} = \sqrt{13} \approx 3.61
$$

$$
||B|| = \sqrt{(7^2) + (1^2)} = \sqrt{49 + 1} = \sqrt{50} \approx 7.07
$$

3. **Step 3: Calculate the Cosine Similarity**:

$$
\text{Cosine Similarity} = \frac{A \cdot B}{||A|| \times ||B||} = \frac{17}{3.61 \times 7.07} \approx \frac{17}{25.53} \approx 0.67
$$

4. **Step 4: Calculate the Cosine Distance**:

$$
\text{Cosine Distance} = 1 - 0.67 = 0.33
$$

Thus, the **Cosine Distance** between \( A(2, 3) \) and \( B(7, 1) \) is approximately **0.33**, indicating that the vectors are somewhat similar in direction.


```
import numpy as np
from scipy.spatial.distance import cosine

# Define the coordinates of points A and B as vectors
A = np.array([2, 3])
B = np.array([7, 1])

# Calculate Cosine Distance using scipy's cosine function
cosine_distance = cosine(A, B)

# Print the result
print(f'Cosine Distance between A and B: {cosine_distance:.2f}')
```
